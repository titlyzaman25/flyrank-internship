## 3. Success metric

*One metric you can defend. What number means 'good'?*
I will use Precision@K because the practical goal is to provide a useful shortlist for human review. For example, Precision@50 measures the proportion of the 50 highest-ranked pages that match the defined positive review proxy. A higher Precision@K means that the limited review capacity is being concentrated on pages that better match the intended review condition.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
A fixed rule could prioritize pages using a single manually chosen condition, such as whether a page is showing a downward trend. However, content opportunity may depend on multiple observable signals at the same time, including search visibility, clicks, CTR, position, engagement, content age, and trends. A ranking model could combine these signals and learn patterns in the data rather than relying on one manually selected threshold. The value of ML would therefore need to be demonstrated by comparing its ranking performance against a simple rule-based baseline using Precision@K.
