## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd

df = pd.read_csv("flyrank-internship/data/raw/content_refresh_anonymized.csv")

print(f"Rows: {len(df):,}")
print(f"Unique pages: {df['content_id'].nunique():,}")
# 1. Number of unique content pages
unique_pages = df["content_id"].nunique()

# 2. Number of pages showing a downward trend
declining_pages = (df["trend_direction"] == "down").sum()

# 3. Number of pages with substantial visibility but low CTR
low_ctr_pages = (
    (df["impressions_90d"] >= 500) &
    (df["ctr"] < 0.5)
).sum()

print(f"Unique content pages: {unique_pages:,}")
print(f"Pages with downward trend: {declining_pages:,}")
print(f"Pages with >=500 impressions and CTR <0.5: {low_ctr_pages:,}")
```
